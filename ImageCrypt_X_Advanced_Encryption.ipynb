```python
import os
import time
from cryptography.hazmat.primitives.ciphers.aead import AESGCM
from cryptography.hazmat.primitives.kdf.hkdf import HKDF
from cryptography.hazmat.primitives import hashes

master_secret = os.urandom(32)
hkdf_salt = os.urandom(16)

hkdf = HKDF(
    algorithm=hashes.SHA256(),
    length=32,
    salt=hkdf_salt,
    info=b"ImageCrypt-X"
)

aes_key = hkdf.derive(master_secret)

print("256-bit AES key generated!")
```

```python
nonce = os.urandom(12)
aesgcm = AESGCM(aes_key)

start_time = time.time()

ciphertext = aesgcm.encrypt(
    nonce,
    image_data,
    None
)

encryption_time = time.time() - start_time

print("Image encrypted successfully!")
print("Original size:", len(image_data), "bytes")
print("Encrypted size:", len(ciphertext), "bytes")
print("Encryption time:", round(encryption_time, 6), "seconds")
```

```python
# Read the encrypted file
with open("ImageCrypt_X_Encrypted.icx", "rb") as file:
    package = file.read()

# Extract the stored components
header = package[:4]
salt = package[4:20]
nonce = package[20:32]
encrypted_data = package[32:]

# Verify file format
assert header == b"ICX1", "Invalid encrypted file!"

# Derive the same AES key
hkdf = HKDF(
    algorithm=hashes.SHA256(),
    length=32,
    salt=salt,
    info=b"ImageCrypt-X"
)

recovered_key = hkdf.derive(master_secret)

# Decrypt the image
decrypted_data = AESGCM(recovered_key).decrypt(
    nonce,
    encrypted_data,
    None
)

print("Image decrypted successfully!")
print("Recovered data size:", len(decrypted_data), "bytes")
```

```python
original_hash = hashlib.sha256(image_data).hexdigest()
recovered_hash = hashlib.sha256(decrypted_data).hexdigest()

print("Original SHA-256:", original_hash)
print("Recovered SHA-256:", recovered_hash)

if original_hash == recovered_hash:
    print("SUCCESS: Original image recovered exactly!")
else:
    print("ERROR: Image recovery mismatch!")
```

```python
# Create a modified copy of the encrypted data
tampered_data = bytearray(encrypted_data)

# Change one byte
tampered_data[0] ^= 1

# Attempt to decrypt the modified data
try:
    AESGCM(recovered_key).decrypt(
        nonce,
        bytes(tampered_data),
        None
    )

    print("Warning: Modified data was accepted!")

except Exception:
    print("SUCCESS: Tampering detected!")
    print("Unauthorized modification rejected.")
```

```python
# Generate an incorrect AES key
wrong_key = os.urandom(32)

# Attempt decryption using the wrong key
try:
    AESGCM(wrong_key).decrypt(
        nonce,
        encrypted_data,
        None
    )

    print("Warning: Incorrect key was accepted!")

except Exception:
    print("SUCCESS: Wrong key rejected!")
    print("Image cannot be decrypted with an incorrect key.")
```
